```python
#Task1
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [5500, 8900, 7100, 87000, 5800, 11500, 9300]

plt.plot(days, steps, marker="o")

plt.title("Daily Steps - Last 7 Days")
plt.xlabel("Day")
plt.ylabel("Number of Steps")

plt.grid(True)

plt.savefig("steps_lineplot.png")

plt.show()
```

```python
#Task2
import matplotlib.pyplot as plt

restaurants = [
    "Dominos", "KFC", "McDonalds", "Subway", "Pizza Hut",
    "Burger King", "Havmor", "La Pinoz", "Honest", "Barbeque Nation"
]

ratings = [4.2, 4.0, 4.3, 4.1, 4.0, 4.2, 4.4, 4.3, 4.1, 4.5]

meal_price = [350, 400, 300, 250, 450, 350, 300, 400, 250, 800]

plt.scatter(ratings, meal_price)

plt.title("Restaurant Ratings vs Average Meal Price")
plt.xlabel("Zomato Rating")
plt.ylabel("Average Meal Price (₹)")

plt.grid(True)

plt.show()
```

```python
#Task3
import matplotlib.pyplot as plt

apps = ["Swiggy", "Zomato", "Domino's"]
orders = [18, 8, 4]

colors = ["blue", "green", "red"]

bars = plt.bar(apps, orders, color=colors)

plt.title("Food Delivery Orders - Last Month")
plt.xlabel("Food Delivery App")
plt.ylabel("Number of Orders")

plt.legend(bars, apps, title="Apps")

plt.show()
```

```python
#Task4
import matplotlib.pyplot as plt

durations = [
    25, 40, 35, 50, 45,
    30, 60, 55, 42, 38,
    70, 65, 32, 48, 52,
    28, 75, 80, 45, 58
]

plt.hist(durations, bins=5, color="purple", edgecolor="black")

plt.title("Spotify Listening Session Durations")
plt.xlabel("Duration (minutes)")
plt.ylabel("Frequency")

plt.show()
```

```python
#Task5
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

screen_time = [60, 75, 55, 90, 80, 120, 100]

posts_liked = [20, 25, 18, 30, 27, 40, 35]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].plot(days, screen_time, marker="o")
axes[0].set_title("Daily Instagram Screen Time")
axes[0].set_xlabel("Day")
axes[0].set_ylabel("Screen Time (minutes)")
axes[0].grid(True)

axes[1].bar(days, posts_liked, color="orange")
axes[1].set_title("Instagram Posts Liked")
axes[1].set_xlabel("Day")
axes[1].set_ylabel("Number of Posts Liked")

plt.tight_layout()

plt.savefig("social_media_usage.png")

plt.show()
```
